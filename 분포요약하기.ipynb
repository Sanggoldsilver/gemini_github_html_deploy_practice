```python
#히스토그램 그리기
#수치형 특성의 값을 일정한 구간(bin)으로 나누어 구간 안에 포함된 데이터 개수를 막대 그래프로 표현

#hist() 함수
#1차원 데이터를 입력받아 히스토그램 그림 -> 기본적으로 10개 구간으로 나눔
#8개 정수로 이루어진 간단한 데이터 사용, bins 매개변수를 5로 지정하여 5개 구간으로 그리기

plt.hist([0, 3, 5, 6, 7, 7, 9, 13], bins=5)
plt.show()
```

```python
#히스토그램의 구간을 정확하게 확인하기 위해 넘파이의 histogram_bin_edges() 함수 사용
import numpy as np
np.histogram_bin_edges([0, 3, 5, 6, 7, 7, 9, 13], bins=5)
```

```python
#randn() 함수는 표준전규분포를 따르는 랜덤한 실수를 생성
np.random.seed(42)
random_samples = np.random.randn(1000)
```

```python
#1000개의 실수로 이루어진 random_samples 배열이 표준정규분포를 따르고 있는지 확인하기 위해 평균과 표준편차를 계산
print(np.mean(random_samples), np.std(random_samples))
```

```python
#히스토그램을 그려서 종 모양의 분포가 나오는지 확인
plt.hist(random_samples)
plt.show()
```

```python
#남산도서관 대출 데이터에서 수치 데이터의 분포 확인
#'대출건수' 열의 히스토그램
plt.hist(ns_book7['대출건수'])
plt.show()
```

```python
#구간 조정하기
#한 구간의 도수가 너무 커서 다른 구간에는 도수가 표시되지 않는 현상이 발생
#y축을 로그 스케일로 바꾸어 해결 - 아래 그림처럼 y축에 로그 함수를 적용
#맷플롯립에서 y축을 로그 스케일로 바꾸려면 yscale() 함수에 'log'를 지정

plt.hist(ns_book7['대출건수'])
plt.yscale('log')
plt.show()
```

```python
#x축의 구간을 더 세세하게 나누기
#hist() 함수는 기본 10개의 구간을 사용
#bins 매개변수에서 이를 100으로 바꿔 데이터 분포를 좀 더 세밀하게 관찰

plt.hist(ns_book7['대출건수'], bins=100)
plt.yscale('log')
plt.show()
```

```python
#'도서명' 열에 apply() 메서드 사용 -> 파이썬 len() 함수 적용 -> title_len 변수는 각 도서명의 길이가 저장된 판다스 객체가 됨
title_len = ns_book7['도서명'].apply(len)
plt.hist(title_len, bins=100)
plt.show()
```

```python
#x축에도 로그 스케일 적용
plt.hist(title_len, bins=100)
plt.xscale('log')
plt.show()
```

```python
#boxplot() 함수로 남산도서관 대출 데이터에서 '대출건수'와 '도서권수' 열의 상자 수염 그림 그리기
plt.boxplot(ns_book7[['대출건수', '도서권수']])
plt.show()
```

```python
plt.boxplot(ns_book7[['대출건수', '도서권수']])
plt.yscale('log')
plt.show()
```

```python
#상자 수염 그림 수평으로 그리기
#boxplot() 함수의 vert 매개변를 기본값 True에서 False로 바꿔줌

plt.boxplot(ns_book7[['대출건수', '도서권수']], vert=False)
plt.xscale('log')
plt.show()
```

```python
#수염 길이 조정하기
#기본적인 수염의 길이는 IQR의 1.5배
#기본값 1.5를 10으로 바꾸어서 IQR의 10배 범위 안에서 가장 멀리 떨어진 데이터까지 수염 그리기
plt.boxplot(ns_book7[['대출건수', '도서권수']], whis=10)
plt.yscale('log')
plt.show()
```

```python
#whis 매개변수는 백분율로도 지정 가능
#(10, 90)으로 지정하면 10%, 90% 백분위수에 해당하는 데이터까지 수염을 그림
#(0, 100)으로 지정하면 마지막 데이터까지 수염을 그리게 됨

plt.boxplot(ns_book7[['대출건수', '도서권수']], whis=(0,100))
plt.yscale('log')
plt.show()
```

```python
#산점도 그리기
#scatter() 메서드
#맷플롯립과 달리 데이터프레임으로 그래프를 그릴 때는 다음처럼 x축과 y축에 해당하는 열 이름만 지정

ns_book7.plot.scatter('도서권수', '대출건수', alpha=0.1)
plt.show()
```
