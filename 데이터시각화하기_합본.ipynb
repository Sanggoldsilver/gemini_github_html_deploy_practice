```python
#맷플롯립 기본 요소 알아보기
#Figuree 객체 - 모든 그래프 구성 요소 담고 있는 최상위 객체

import gdown
gdown.download('https://bit.ly/3pK7iuu', 'ns_book7.csv', quiet=False)

import pandas as pd

ns_book7 = pd.read_csv('ns_book7.csv', low_memory=False)
ns_book7.head()
```

```python
#산점도 그리기
#맷플롯립 임포트 -> scatter() -> 도서권수 x 대출건수 y 투명도 0.1

import matplotlib.pyplot as plt

plt.scatter(ns_book7['도서권수'], ns_book7['대출건수'], alpha=0.1)
plt.show()
```

```python
#그래프 크기 바꾸기 : figsize 매개변수
#figsize 매개변수에 그래프의 크기를 튜플로 지정
# - 기본 그래프 크기는 (6,4), 각각 너비와 높이는 inch 단위
#-> (9,6)으로 지정 -> 산점도 좀 더 크게 그리기

plt.figure(figsize=(9,6)) #(너비,높이)
plt.scatter(ns_book7['도서권수'], ns_book7['대출건수'], alpha=0.1)
plt.show()
```

```python
#subplots()함수
#함수에 원하는 서브 플롯 개수 지정
#subplots() 함수로 두 개의 서브플롯을 그리고 싶다면 매개변수에 2를 입력
#이 배열의 원소에서 각각 scatter() 함수와 hist() 함수 호출
#첫 번째 그래프인 산점도의 x축은 '도서권수' 열, y축은 '대출건수'열
#두 번째 그래프는 히스토그램, X축은 '대출건수'열, y축은 '대출건수의 빈도'
#히스토그램의 y축은 로그 스케일로 지정

fig, axs = plt.subplots(2)

axs[0].scatter(ns_book7['도서권수'], ns_book7['대출건수'], alpha=0.1)
#axs[0] 이게 첫 번째 그래프를 의미

axs[1].hist(ns_book7['대출건수'], bins=100)
axs[1].set_yscale('log')

fig.show()
```

* 산점도와 히스토그램이 두 개의 그래프로 위아래 나뉘어져 그려짐


* 피겨 크기의 기본값인 (6, 4) 안에서 두 개의 서브플롯을 배치하기 때문에 각 그래프의 높이가 많이 줄었음

```python
#subplots() 함수도 피겨 크기를 지정할 수 있는 figsize 매개변수 제공
#(6, 8)로 지정 및 높이 조절
#또한 set_title() 메서드 사용 -> 각 그래프에 제목 삽입

fig, axs = plt.subplots(2, figsize=(6, 8))

axs[0].scatter(ns_book7['도서권수'], ns_book7['대출건수'], alpha=0.1)
axs[0].set_title('scatter plot')
#set_title은 서브플롯의 제목임

axs[1].hist(ns_book7['대출건수'], bins=100)
axs[1].set_title('histogram')
axs[1].set_yscale('log')

fig.show()
```

```python
import gdown
gdown.download('https://bit.ly/3pK7iuu', 'ns_book7.csv', quiet=False)
```

```python
import pandas as pd

ns_book7 = pd.read_csv('ns_book7.csv', low_memory=False)
ns_book7.head()
```

```python
#value_counts() 메서드로 연도별 도서 개수 구하기
count_by_year = ns_book7['발행년도'].value_counts()#고유한 값의 등장 횟수 게산
count_by_year
```

```python
count_by_year = count_by_year.sort_index()#인덱스 별로 정리
count_by_year
```

```python
count_by_year = count_by_year[count_by_year.index <= 2030] #2030년보다 작거나 같은
count_by_year
```

image.png

```python
import numpy as np

def kdc_1st_char(no):
    if no is np.nan:
        return '-1'
    else:
        return no[0]

count_by_subject = ns_book7['주제분류번호'].apply(kdc_1st_char).value_counts()
count_by_subject
```

image.png

```python
#선 그래프 그리기
#plot() 함수 - 첫 번재 매개변수 X축 값, 두 번재 매개변수 y축 해당 값 전달
#해상도 높이기 위해 기본 DPI 100으로 변경

import matplotlib.pyplot as plt
plt.rcParams['figure.dpi'] = 100

#plot() 함수의 두 매개변수에 count_by_year의 연도(index)와 도서 개수(values)를 각각 지정
plt.plot(count_by_year.index, count_by_year.values)
plt.title('Books by year')
plt.xlabel('year')
plt.ylabel('number of books')
plt.show()
```

image.png

```python
plt.plot(count_by_year, marker='.', linestyle=':', color='red')
plt.title('Books by year')
plt.xlabel('year')
plt.ylabel('number of books')
plt.show()
```

```python
#plot() 함수의 매개 변수에 지정한 마커, 선 모양, 색갈을 하나의 문자열로 합쳐서 지정
plt.plot(count_by_year, '.:r') #<- 마커, 선 모양, 색깔 옵션을 하나의 문자열로 지정 가능
```

```python
#별 모양 마커와 실선을 사용한 녹색 선 그래프
plt.plot(count_by_year, '*-g')
plt.title('Books by year')
plt.xlabel('year')
plt.ylabel('number of books')
plt.show()
```

image.png

```python
#선 그래프 눈금 개수 조절 및 마커에 텍스트 표시하기
#xticks() 함수 - x축 눈금 지정
#range() 함수 - 1947년부터 2030년까지 10년씩 건너뒤면서 x축 눈금 표시
#슬라이스 연산자(:) - 다섯 개씩 건너뛰면서 count_by_year의 값 선택
#items() 메서드 - 인덱스와 값을 감싼 튜플 얻기
#annotate() 함수 - 그래프에 값을 표시

plt.plot(count_by_year, '*-g')
plt.title('Books by year')
plt.xlabel('year')
plt.ylabel('number of books')
plt.xticks(range(1947, 2030, 10))
for idx, val in count_by_year[::5].items():
  plt.annotate(val, (idx, val))
plt.show()
```

```python
#xytext 매개변수 - 텍스트 위치 조절
#annotate() 함수는 기본적으로 마커와 동일한 좌표 시스템 사용
#다음처럼 마커 위치에서 x축은 1만큼, y축으로는 10만큼 떨어지게 지정

plt.plot(count_by_year, '*-g')
plt.title('Books by year')
plt.xlabel('year')
plt.ylabel('number of books')
plt.xticks(range(1947, 2030, 10))
for idx, val in count_by_year[::5].items():
  plt.annotate(val, (idx, val), xytext=(idx+1, val+10))
plt.show()
```

```python
#xytext 매개변수와 함께 textcoords 매개변수 사용 - 상대적인 위치를 포인트나 픽셀 단위로 지정
plt.plot(count_by_year, '*-g')
plt.title('Books by year')
plt.xlabel('year')
plt.ylabel('number of books')
plt.xticks(range(1947, 2030, 10))
for idx, val in count_by_year[::5].items():
  plt.annotate(val, (idx, val), xytext=(2, 2), textcoords='offset points')
plt.show()
```

```python
#bar() 함수 - x축의 값과 막대 높이에 해당하는 y축의 값을 전달

plt.bar(count_by_subject.index, count_by_subject.values)
plt.xlabel('subject')
plt.ylabel('number of books')
for idx, val in count_by_subject.items():
  plt.annotate(val, (idx, val), xytext=(0, 2), textcoords='offset points')
plt.show()
```
